```r
### EDIT THIS CODE CELL ##

sum_stat<- NCAA|>
    summarise(across(where(is.numeric), #apply the summarise across all columns that are numeric
                    list(mean = ~mean(.x, na.rm = TRUE), #finds mean across all columns and removes missing values
                        min = ~min(.x, na.rm = TRUE), #finds minimum across all columns and removes missing values
                        max = ~max(.x, na.rm = TRUE))))#finds maximum across all columns and removes missing values
sum_stat
```

```r
### EDIT THIS CODE CELL FOR VISUALIZATION 1###

# Can a school team's seed, regional win percentage, total number of games, and total win percentage be 
# used to predict where they will advance in the tournament?

avg_tot_win_vs_seed_plot<- NCAA|>
    group_by(seed)|> 
    filter(seed != 0)|>
    summarize(mean_win = mean(total_wins_pct))|>
    ggplot(aes(x = seed, y = mean_win))+
    geom_bar(stat = "identity")+ 
    labs(x = "Seed", y = "Average Win Percentage of Seed", title = "Average Total Winning Percentage vs Seed")+ 
    theme(text = element_text(size = 12))
avg_tot_win_vs_seed_plot
```
